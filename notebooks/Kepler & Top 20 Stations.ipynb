```python
import pandas as pd
from pathlib import Path

RAW_DIR = Path("data_raw/csv")
files = sorted(list(RAW_DIR.glob("*.csv"))) + sorted(list(RAW_DIR.glob("*.csv.gz")))

keep_cols = [
    "start_station_name","end_station_name",
    "start_lat","start_lng","end_lat","end_lng"
]

parts = []
for f in files:
    temp = pd.read_csv(f, usecols=keep_cols, low_memory=False)
    parts.append(temp)

trips_small = pd.concat(parts, ignore_index=True)

od_counts = (
    trips_small.dropna()
    .groupby(keep_cols)
    .size()
    .reset_index(name="trip_count")
)

TOP_N = 2000
od_top = od_counts.sort_values("trip_count", ascending=False).head(TOP_N).copy()

od_top.shape
```

```python
from keplergl import KeplerGl

map_1 = KeplerGl(height=650)
map_1.add_data(data=od_top, name="OD Trips")
map_1
```

```python
map_1.save_to_html(file_name="kepler_od_map.html", read_only=False)
print("Saved kepler_od_map.html")
```
