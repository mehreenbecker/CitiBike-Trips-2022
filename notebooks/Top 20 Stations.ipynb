```python
import pandas as pd
from pathlib import Path

RAW_DIR = Path("data_raw/csv")
files = sorted(list(RAW_DIR.glob("*.csv"))) + sorted(list(RAW_DIR.glob("*.csv.gz")))

counts = {}

for f in files:
    temp = pd.read_csv(f, usecols=["start_station_name"], low_memory=False)
    vc = temp["start_station_name"].dropna().value_counts()
    for k, v in vc.items():
        counts[k] = counts.get(k, 0) + int(v)

top20 = (
    pd.Series(counts)
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

top20.columns = ["start_station_name", "trip_count"]

top20.head()
```

```python
out_path = Path("data_processed/top_20_stations.csv")
top20.to_csv(out_path, index=False)

print("Saved to:", out_path)
```
